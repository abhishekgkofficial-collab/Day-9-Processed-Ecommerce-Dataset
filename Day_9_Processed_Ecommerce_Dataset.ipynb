```python
# ============================================================
# DAY 9 INTERNSHIP ASSIGNMENT
# PROCESSED E-COMMERCE DATASET
# Python + Pandas
# ============================================================

import pandas as pd
import numpy as np
from google.colab import files

print("=" * 70)
print("DAY 9 - PROCESSED E-COMMERCE DATASET")
print("=" * 70)

# ============================================================
# 1. UPLOAD THREE CSV FILES
# ============================================================

print("\nPlease upload these 3 files:")
print("1. Orders CSV")
print("2. Customers CSV")
print("3. Products CSV")

uploaded = files.upload()

print("\nFiles uploaded successfully!")

# Identify files automatically
orders_file = None
customers_file = None
products_file = None

for file_name in uploaded.keys():

    if "order" in file_name.lower():
        orders_file = file_name

    elif "customer" in file_name.lower():
        customers_file = file_name

    elif "product" in file_name.lower():
        products_file = file_name

print("\nDetected Files:")
print("Orders   :", orders_file)
print("Customers:", customers_file)
print("Products :", products_file)


# ============================================================
# 2. LOAD DATASETS
# ============================================================

orders = pd.read_csv(orders_file)
customers = pd.read_csv(customers_file)
products = pd.read_csv(products_file)

print("\n" + "=" * 70)
print("2. DATASETS LOADED")
print("=" * 70)

print("\nOrders Shape:", orders.shape)
print("Customers Shape:", customers.shape)
print("Products Shape:", products.shape)


# ============================================================
# 3. DISPLAY DATASETS
# ============================================================

print("\n" + "=" * 70)
print("3. DATASET PREVIEW")
print("=" * 70)

print("\nORDERS:")
display(orders.head())

print("\nCUSTOMERS:")
display(customers.head())

print("\nPRODUCTS:")
display(products.head())


# ============================================================
# 4. CHECK DATA TYPES
# ============================================================

print("\n" + "=" * 70)
print("4. DATA TYPES")
print("=" * 70)

print("\nOrders:")
print(orders.dtypes)

print("\nCustomers:")
print(customers.dtypes)

print("\nProducts:")
print(products.dtypes)


# ============================================================
# 5. CHECK MISSING VALUES
# ============================================================

print("\n" + "=" * 70)
print("5. MISSING VALUES")
print("=" * 70)

print("\nOrders Missing Values:")
print(orders.isnull().sum())

print("\nCustomers Missing Values:")
print(customers.isnull().sum())

print("\nProducts Missing Values:")
print(products.isnull().sum())


# ============================================================
# 6. MERGE ORDERS WITH CUSTOMERS
# ============================================================

print("\n" + "=" * 70)
print("6. MERGING ORDERS AND CUSTOMERS")
print("=" * 70)

orders_customers = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)

print("Orders + Customers Shape:",
      orders_customers.shape)

display(orders_customers.head())


# ============================================================
# 7. MERGE WITH PRODUCTS
# ============================================================

print("\n" + "=" * 70)
print("7. MERGING WITH PRODUCTS")
print("=" * 70)

processed_df = pd.merge(
    orders_customers,
    products,
    on="Product_ID",
    how="left"
)

print("Final Merged Shape:",
      processed_df.shape)

display(processed_df.head())


# ============================================================
# 8. CONCAT DEMONSTRATION
# ============================================================

print("\n" + "=" * 70)
print("8. CONCAT DEMONSTRATION")
print("=" * 70)

# Split Orders into two parts
orders_part1 = orders.iloc[:60]
orders_part2 = orders.iloc[60:]

# Combine them using concat()
orders_combined = pd.concat(
    [orders_part1, orders_part2],
    ignore_index=True
)

print("First Part Shape:", orders_part1.shape)
print("Second Part Shape:", orders_part2.shape)
print("Combined Shape:", orders_combined.shape)

print("\nCombined DataFrame:")
display(orders_combined.head())


# ============================================================
# 9. DATETIME CONVERSION
# ============================================================

print("\n" + "=" * 70)
print("9. DATETIME PROCESSING")
print("=" * 70)

processed_df["Order_Date"] = pd.to_datetime(
    processed_df["Order_Date"]
)

print("Order_Date converted to:")
print(processed_df["Order_Date"].dtype)


# ============================================================
# 10. EXTRACT DATE INFORMATION
# ============================================================

print("\n" + "=" * 70)
print("10. EXTRACTING DATE INFORMATION")
print("=" * 70)

processed_df["Order_Year"] = (
    processed_df["Order_Date"].dt.year
)

processed_df["Order_Month"] = (
    processed_df["Order_Date"].dt.month
)

processed_df["Order_Month_Name"] = (
    processed_df["Order_Date"].dt.month_name()
)

processed_df["Order_Day"] = (
    processed_df["Order_Date"].dt.day
)

processed_df["Order_Day_Name"] = (
    processed_df["Order_Date"].dt.day_name()
)

display(
    processed_df[
        [
            "Order_Date",
            "Order_Year",
            "Order_Month",
            "Order_Month_Name",
            "Order_Day",
            "Order_Day_Name"
        ]
    ].head(10)
)


# ============================================================
# 11. APPLY() - CREATE TOTAL SALES
# ============================================================

print("\n" + "=" * 70)
print("11. APPLY() - CALCULATING TOTAL SALES")
print("=" * 70)

processed_df["Total_Sales"] = processed_df.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)

display(
    processed_df[
        [
            "Order_ID",
            "Product_Name",
            "Quantity",
            "Unit_Price",
            "Total_Sales"
        ]
    ].head(10)
)


# ============================================================
# 12. APPLY() - CREATE ORDER VALUE CATEGORY
# ============================================================

print("\n" + "=" * 70)
print("12. APPLY() - ORDER VALUE CATEGORY")
print("=" * 70)

def classify_order(value):

    if value >= 3000:
        return "High Value"

    elif value >= 1500:
        return "Medium Value"

    else:
        return "Low Value"


processed_df["Order_Value_Category"] = (
    processed_df["Total_Sales"].apply(classify_order)
)

display(
    processed_df[
        [
            "Order_ID",
            "Total_Sales",
            "Order_Value_Category"
        ]
    ].head(10)
)


# ============================================================
# 13. CLEAN FINAL DATAFRAME
# ============================================================

print("\n" + "=" * 70)
print("13. CLEAN PROCESSED DATASET")
print("=" * 70)

final_columns = [
    "Order_ID",
    "Order_Date",
    "Order_Year",
    "Order_Month",
    "Order_Month_Name",
    "Order_Day",
    "Order_Day_Name",
    "Customer_ID",
    "Customer_Name",
    "City",
    "Region",
    "Membership_Type",
    "Product_ID",
    "Product_Name",
    "Category",
    "Brand",
    "Unit_Price",
    "Quantity",
    "Total_Sales",
    "Order_Value_Category",
    "Payment_Method",
    "Order_Status"
]

final_df = processed_df[final_columns].copy()

print("Final Dataset Shape:")
print(final_df.shape)

display(final_df.head(10))


# ============================================================
# 14. BASIC ANALYSIS OF PROCESSED DATA
# ============================================================

print("\n" + "=" * 70)
print("14. PROCESSED DATASET ANALYSIS")
print("=" * 70)

print("\nTotal Orders:",
      final_df["Order_ID"].nunique())

print("Total Quantity Sold:",
      final_df["Quantity"].sum())

print("Total Sales:",
      final_df["Total_Sales"].sum())

print("Average Order Value:",
      final_df["Total_Sales"].mean())

print("Highest Order Value:",
      final_df["Total_Sales"].max())

print("Lowest Order Value:",
      final_df["Total_Sales"].min())


# ============================================================
# 15. SALES BY CATEGORY
# ============================================================

print("\n" + "=" * 70)
print("15. SALES BY CATEGORY")
print("=" * 70)

category_sales = (
    final_df
    .groupby("Category")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

display(category_sales)


# ============================================================
# 16. SALES BY CITY
# ============================================================

print("\n" + "=" * 70)
print("16. SALES BY CITY")
print("=" * 70)

city_sales = (
    final_df
    .groupby("City")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

display(city_sales)


# ============================================================
# 17. SALES BY PRODUCT
# ============================================================

print("\n" + "=" * 70)
print("17. SALES BY PRODUCT")
print("=" * 70)

product_sales = (
    final_df
    .groupby("Product_Name")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

display(product_sales.head(10))


# ============================================================
# 18. SALES BY PAYMENT METHOD
# ============================================================

print("\n" + "=" * 70)
print("18. SALES BY PAYMENT METHOD")
print("=" * 70)

payment_sales = (
    final_df
    .groupby("Payment_Method")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
)

display(payment_sales)


# ============================================================
# 19. SALES BY MONTH
# ============================================================

print("\n" + "=" * 70)
print("19. MONTHLY SALES")
print("=" * 70)

monthly_sales = (
    final_df
    .groupby("Order_Month_Name")["Total_Sales"]
    .sum()
)

display(monthly_sales)


# ============================================================
# 20. EXPORT FINAL DATASET
# ============================================================

print("\n" + "=" * 70)
print("20. EXPORTING FINAL DATASET")
print("=" * 70)

output_file = "Day9_Processed_Ecommerce_Dataset.csv"

final_df.to_csv(
    output_file,
    index=False
)

print("Processed dataset saved as:")
print(output_file)


# ============================================================
# 21. DOWNLOAD PROCESSED DATASET
# ============================================================

files.download(output_file)


# ============================================================
# 22. FINAL OBSERVATIONS
# ============================================================

print("\n" + "=" * 70)
print("22. FINAL OBSERVATIONS")
print("=" * 70)

best_category = (
    final_df.groupby("Category")["Total_Sales"]
    .sum()
    .idxmax()
)

best_city = (
    final_df.groupby("City")["Total_Sales"]
    .sum()
    .idxmax()
)

best_product = (
    final_df.groupby("Product_Name")["Total_Sales"]
    .sum()
    .idxmax()
)

best_payment = (
    final_df.groupby("Payment_Method")["Total_Sales"]
    .sum()
    .idxmax()
)

print("Best Performing Category:", best_category)
print("Best Performing City:", best_city)
print("Best Performing Product:", best_product)
print("Highest Sales Payment Method:", best_payment)

print("\nTotal Sales:",
      final_df["Total_Sales"].sum())

print("Total Quantity Sold:",
      final_df["Quantity"].sum())

print("Number of Orders:",
      final_df["Order_ID"].nunique())


# ============================================================
# COMPLETED
# ============================================================

print("\n" + "=" * 70)
print("DAY 9 ASSIGNMENT COMPLETED SUCCESSFULLY")
print("=" * 70)
```
